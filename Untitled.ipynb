### The 64 borrower profiles form a broadly ordered risk gradient, with increasingly elevated default rates and some larger separations toward the upper-risk end.

```python
from sklearn.model_selection import train_test_split

train_data, test_data = train_test_split(
    loan_data,
    test_size=0.30,
    random_state=42,
    stratify=loan_data['Default']
)

print("Train:", len(train_data))
print("Test:", len(test_data))

print("Train default rate:", train_data['Default'].mean() * 100)
print("Test default rate:", test_data['Default'].mean() * 100)
```

```python
train_profiles = (
    train_data.groupby(
        ['Interest_bands', 'EmploymentType', 'DTI_band'],
        observed=False
    )
    .agg(
        Borrowers=('LoanID', 'count'),
        defaults=('Default', 'sum')
    )
    .reset_index()
)

train_profiles['default_rate'] = (
    train_profiles['defaults'] /
    train_profiles['Borrowers'] * 100
).round(2)
```

```python
train_profiles['risk_index'] = (
    train_profiles['default_rate'] / 11.61
).round(2)
```

```python
train_profiles[
    ['Interest_bands',
     'EmploymentType',
     'DTI_band',
     'Borrowers',
     'default_rate',
     'risk_index']
].sort_values('risk_index')
```

```python
train_profiles['risk_zone'] = pd.cut(
    train_profiles['risk_index'],
    bins=[0, 0.75, 1.00, 1.30, 1.60, float('inf')],
    labels=[
        'Very Low',
        'Below Average',
        'Around Average',
        'Elevated',
        'Very Elevated'
    ]
)
```

```python
risk_map = train_profiles[
    ['Interest_bands', 'EmploymentType', 'DTI_band', 'risk_zone']
]
```

```python
test_segmented = test_data.merge(
    risk_map,
    on=['Interest_bands', 'EmploymentType', 'DTI_band'],
    how='left'
)
```

```python
test_summary = (
    test_segmented
    .groupby('risk_zone', observed=False)
    .agg(
        Borrowers=('LoanID', 'count'),
        defaults=('Default', 'sum')
    )
    .reset_index()
)

test_summary['default_rate'] = (
    test_summary['defaults'] /
    test_summary['Borrowers'] * 100
).round(2)

test_summary['portfolio_share'] = (
    test_summary['Borrowers'] /
    test_summary['Borrowers'].sum() * 100
).round(2)
```

## Final Business Insights

### 1. Risk is not explained by a single borrower characteristic
The analysis shows that credit risk is better understood through a combination of financial characteristics rather than one demographic variable.

### 2. Interest rate provides a meaningful risk signal
Default rates vary across interest-rate bands, indicating that pricing and borrower risk are related.

### 3. DTI provides additional financial-risk context
Higher debt-to-income levels can indicate greater repayment pressure and become more informative when combined with other risk factors.

### 4. Employment type alone is not sufficient
The very-elevated-risk segment contains borrowers across all employment categories in relatively similar proportions. Therefore, employment status should not be used as a standalone reason to classify a borrower as high risk.

### 5. Dependents are not a standalone risk indicator
The very-elevated-risk segment is almost evenly split between borrowers with and without dependents.

### Business takeaway
Credit risk should be assessed using a combination of financial characteristics rather than relying on a single borrower attribute. The segmentation analysis provides a more defensible way to identify groups with substantially different observed default rates.

### Conclusion

* This analysis developed an interpretable borrower risk-segmentation framework using interest-rate band, employment type and DTI band. The resulting 64 borrower profiles demonstrated substantial variation in observed default rates, ranging from 4.70% to 21.74%, compared with an overall portfolio default rate of 11.61%.

* Interest rate emerged as the strongest risk differentiator, while employment type provided substantial additional separation. DTI contributed further refinement, although its impact varied across different interest-rate and employment combinations.

* The profiles were consolidated into five practical risk zones. On a held-out test sample, observed default rates increased consistently from 6.90% in the Very Low zone to 19.51% in the Very Elevated zone, demonstrating that the segmentation retained its risk ordering on unseen observations.

* The highest-risk segment was particularly concentrated among unemployed borrowers, while characteristics such as loan purpose, education, mortgage status and dependents showed limited differentiation within that segment.

* From a business perspective, the segmentation can support differentiated underwriting intensity, monitoring and pricing review rather than treating all borrowers identically. Higher observed risk should trigger stronger credit controls rather than automatically result in rejection.

*The analysis is descriptive and segmentation-based rather than a production probability-of-default model. Further work would be required to incorporate profitability, loss severity, calibration, model validation and out-of-time testing before operational deployment.*
